-----------

ENCODING

```python
from sklearn.preprocessing import LabelEncoder
```

```python
feature_cols = ["AnimalName",	"symptoms1",	"symptoms2",	"symptoms3",	"symptoms4",	"symptoms5"]

for cols in feature_cols:
    le = LabelEncoder()
    df[cols] = le.fit_transform(df[cols])
```

```python
df
```

```python
df['Dangerous'] = le.fit_transform(df['Dangerous'])
```

```python
df
```

----

TRAIN_TEST_SPLIT

```python
x = df.drop(columns = "Dangerous")
x
```

```python
y = df["Dangerous"]
y
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,random_state = 99, stratify = y, test_size=0.3)
```

---

TRAINING AND TESTING

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
```

```python
model = LogisticRegression()
```

```python
model
```

```python
model.fit(x_train ,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
accuracy_score(y_pred, y_test)
```

```python
cmap = confusion_matrix(y_test, y_pred)
sns.heatmap(cmap,annot= True)
```
